1. Pandas Basic

**คำอธิบาย:** Pandas เป็นไลบรารีหลักที่ใช้จัดการข้อมูลเชิงตาราง (tabular data) ใน Python โดยโครงสร้างข้อมูลพื้นฐานที่สุดมี 2 แบบ:
1. **Series** - โครงสร้างข้อมูล 1 มิติ เหมือนคอลัมน์เดียวในตาราง หรืออาร์เรย์ที่มี index เป็นของตัวเอง
2. **DataFrame** - โครงสร้างข้อมูล 2 มิติ เหมือนตาราง Excel ที่มีทั้งแถวและคอลัมน์ สามารถเก็บข้อมูลหลายชนิดได้ในเวลาเดียวกัน

เซลล์นี้แสดงวิธีสร้าง **Series** จาก 3 แหล่งข้อมูล ได้แก่ List, Dictionary และไฟล์ CSV การเข้าถึงค่าทำได้ทั้งแบบดูทั้งก้อนข้อมูล (`print(s)`) หรือดูค่าตำแหน่งเดียวด้วย `.iloc[]` (อ้างอิงตำแหน่งตัวเลข) หรือใช้ label ตรง ๆ เช่น `s["Bob"]` เมื่อ index เป็นข้อความ

```python
### Creating Series ###

## Example 1: From List
scores = [85, 90, 78, 92]
s = pd.Series(scores)
print(s)                                            # Access entire series
print("The first element is", s.iloc[3])                 # Access single element

## Example 2: From Dictionary
scores = {
    "Alice": 85,
    "Bob": 90,
    "Charlie": 78
}
s = pd.Series(scores)
print(s)                                            # Access entire series
print("The second element is", s["Bob"])             # Access single element


## Example 3: From .csv (or .xlsx)
s = pd.read_csv("data_series.csv")["S1"]         
print(s)
print(type(s))
```

**คำอธิบาย:** เซลล์นี้แสดงวิธีสร้าง **DataFrame** จาก 3 แหล่งข้อมูล เช่นเดียวกับ Series ได้แก่ List (ระบุ `columns` แยกต่างหาก), Dictionary (แต่ละ key กลายเป็นชื่อคอลัมน์โดยอัตโนมัติ) และไฟล์ CSV ผ่าน `pd.read_csv()`

จะเห็นว่า DataFrame ที่อ่านจากไฟล์ CSV สามารถมีคอลัมน์หลายชนิดข้อมูลปะปนกันได้ในตารางเดียว (ตัวเลขทศนิยม, จำนวนเต็ม, ข้อความ) ซึ่งเป็นจุดเด่นของ DataFrame เมื่อเทียบกับ Series ที่เก็บได้เพียงคอลัมน์เดียว

```python
### Creating DataFrames ###

## Example 1: From List
data = [
    ["Alice", 25, 85],
    ["Bob", 30, 90],
    ["Charlie", 22, 78]
]
df = pd.DataFrame(
    data,
    columns=["Name", "Age", "Score"]
)
print(df)

## Example 2: From Dictionary
data = {
    "Name": ["Brad", "Cuper", "Lisa"],
    "Age": [25, 30, 22],
    "Score": [85, 90, 78]
}
df = pd.DataFrame(data)
print(df)


## Example 3: From .csv (or .xlsx)
df = pd.read_csv("data_dataframe.csv")         
print(df)
print(type(df))
```

**คำอธิบาย:** หลังโหลดข้อมูลเข้ามาแล้ว ควรตรวจสอบข้อมูลเบื้องต้นก่อนนำไปวิเคราะห์ต่อเสมอ โดยมีเมธอดหลักที่ใช้บ่อย ได้แก่
- `df.head()` - ดูข้อมูล 5 แถวแรก (default)
- `df.tail()` - ดูข้อมูล 5 แถวสุดท้าย
- `df.info()` - ดูภาพรวมโครงสร้าง เช่น จำนวนแถว คอลัมน์ ชนิดข้อมูล และค่าที่ไม่ใช่ null ของแต่ละคอลัมน์
- `df.describe()` - ดูค่าสถิติเชิงพรรณนาของคอลัมน์ตัวเลข เช่น mean, std, min, max, quartile

ในตัวอย่างนี้ใช้ `df.head()` เพื่อดูตัวอย่างข้อมูล iris 5 แถวแรก ส่วนบรรทัดอื่นถูก comment ไว้ให้ลองเปิดใช้เองได้

2. Data Selection & Indexing

**คำอธิบาย:** การเลือกคอลัมน์ใน DataFrame ทำได้ 2 แบบที่ให้ผลลัพธ์ต่างกัน:
- `df['column']` (single bracket) - คืนค่าเป็น **Series** (1 มิติ)
- `df[['column1', 'column2']]` (double bracket) - คืนค่าเป็น **DataFrame** (2 มิติ) แม้จะเลือกเพียงคอลัมน์เดียวก็ตาม

ตัวอย่างนี้เลือก 2 คอลัมน์ (`sepal_width`, `sepal_length`) พร้อมกันโดยส่ง list ของชื่อคอลัมน์เข้าไปใน double bracket และลำดับคอลัมน์ในผลลัพธ์จะเรียงตามลำดับที่ระบุใน list ไม่ใช่ลำดับเดิมในไฟล์

```python
### Selecting columns ###

df = pd.read_csv("iris.csv") 
# df['sepal_width']
# df['sepal_length']

df[['sepal_width', 'sepal_length']]
# Take Note
# df['column']                 >>> Series
# df[['column1', 'column2']]   >>> DataFrame
```

**คำอธิบาย:** การเลือกแถวข้อมูลมี 2 เครื่องมือหลักที่ควรแยกให้ออกจากกัน:
- **`.iloc[]`** - เลือกโดยอ้างอิง **ตำแหน่งตัวเลข** (integer position) เหมือน index ของ list/array เช่น `df.iloc[0:5, 0:3]` หมายถึงแถวที่ 0-4 และคอลัมน์ที่ 0-2
- **`.loc[]`** - เลือกโดยอ้างอิงจาก **label** ของ index/คอลัมน์ เช่น หากตั้ง index เป็นตัวอักษร (`"A", "B", "C"`) ต้องใช้ label เหล่านั้นตรง ๆ ใน `.loc[]`

ข้อควรระวัง: ถ้า index ของ DataFrame เป็นตัวเลข (เช่นค่า default 0, 1, 2, ...) `.loc[0]` กับ `.iloc[0]` อาจให้ผลลัพธ์เหมือนกัน แต่ความหมายต่างกันโดยพื้นฐาน - `.loc` ยึด label เป็นหลัก ส่วน `.iloc` ยึดตำแหน่งเป็นหลักเสมอ

```python
### Row selection ###

## Example 1: .iloc[]
df = pd.read_csv("iris.csv") 
df.iloc[0]
df.iloc[0, 1]
df.iloc[[0],[1]]
df.iloc[0:5, 0:3]
df.iloc[5:10, 3:5]


## Example 2: .loc[]
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 22, 28],
    "Score": [85, 90, 78, 92]
}, index=["A", "B", "C", "D"])
print(df)
print()
print(df.loc[["A"]])
print()
print(df.loc["A", "Name"])
print()
print(df.loc["B"]["Score"])                         # OK but not good coding style


## Example 2.1: .loc[] with number index
df = pd.read_csv("iris.csv") 
print(df.loc[0,"sepal_width"])
print(df.loc[0]["sepal_width"])                     # OK but not good coding style
```

**คำอธิบาย:** Boolean filtering คือการกรองแถวข้อมูลด้วยเงื่อนไขที่ให้ผลลัพธ์เป็น True/False เช่น `df["sepal_length"] > 6` จะสร้าง Series ของค่า True/False ตามแต่ละแถว แล้วนำไปใส่ใน `df[...]` เพื่อกรองเฉพาะแถวที่เป็น True

การรวมหลายเงื่อนไขต้องใช้ตัวดำเนินการของ pandas โดยเฉพาะ (ไม่ใช่ `and`/`or` แบบ Python ปกติ):
- `&` แทน AND (ต้องเป็นจริงทุกเงื่อนไข)
- `|` แทน OR (เป็นจริงเงื่อนไขใดเงื่อนไขหนึ่งก็พอ)

แต่ละเงื่อนไขต้องอยู่ในวงเล็บของตัวเองเสมอ เช่น `(df["a"] > 1) & (df["b"] < 2)` นอกจากนี้ยังมีเมธอดช่วยกรองที่สะดวก เช่น `.between()` (กรองช่วงค่า), `.isin()` (กรองว่าค่าตรงกับรายการที่กำหนดหรือไม่) และสามารถใช้ `df.loc[condition, columns]` เพื่อกรองแถวพร้อมเลือกคอลัมน์ที่ต้องการแสดงในคำสั่งเดียว

```python
### Boolean filtering ###

df = pd.read_csv("iris.csv") 
df[df["sepal_length"] > 6]
df[df["sepal_width"] < 3]
df[(df["sepal_length"] > 7) & (df["petal_width"] < 2)]
df[(df["sepal_length"] > 7) | (df["petal_width"] > 2)]
df[df["sepal_length"].between(5, 6)]
df[df["iris_class"] == "Iris-setosa"]
df[df["iris_class"].isin(["Iris-setosa", "Iris-versicolor"])]
df.loc[df["sepal_length"] > 6, ["sepal_length", "sepal_width", "iris_class"]]
```

**คำอธิบาย:** โดย default DataFrame จะมี index เป็นตัวเลขเริ่มจาก 0 ให้อัตโนมัติ แต่บางครั้งเราต้องการใช้คอลัมน์ที่มีความหมาย (เช่น รหัสประจำตัว) เป็น index แทน เพื่อให้ค้นหาแถวด้วย `.loc[]` ได้สะดวกขึ้น

- `df.set_index("ID")` - ตั้งคอลัมน์ `ID` เป็น index ใหม่ (คอลัมน์นั้นจะไม่ปรากฏเป็นคอลัมน์ข้อมูลอีกต่อไป)
- `df.reset_index()` - คืนค่า index กลับมาเป็นตัวเลขปกติ (0, 1, 2, ...) โดยนำ index เดิมกลับมาเป็นคอลัมน์ข้อมูลแทน

ทั้งสองเมธอดนี้คืนค่าเป็น DataFrame ใหม่ (ไม่แก้ไข DataFrame เดิม) เว้นแต่จะระบุ `inplace=True`

```python
### Setting index ###

df = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 21, 19, 22],
    "Score": [85, 90, 78, 92]
})
print(df)

df2 = df.set_index("ID")
print(df2)

df2 = df.reset_index()
print(df2)
```

3. Data Cleaning


**คำอธิบาย:** ข้อมูลจริงมักมีค่าที่ขาดหายไป (missing values หรือ `NaN`) ซึ่ง pandas มีเครื่องมือจัดการหลายแบบ:
- `df.isna()` - ตรวจสอบทีละค่าว่าเป็นค่าว่างหรือไม่ (คืนค่า True/False) ต่อด้วย `.sum()` เพื่อนับจำนวนค่าว่างในแต่ละคอลัมน์
- `df.dropna()` - ลบแถวที่มีค่าว่างออกทั้งหมด หรือระบุ `subset=["col"]` เพื่อลบเฉพาะแถวที่คอลัมน์นั้นว่าง
- `df.fillna(value)` - เติมค่าว่างด้วยค่าที่กำหนด เช่นเติมด้วย 0, ค่าเฉลี่ย (`.mean()`) หรือกำหนดค่าต่างกันในแต่ละคอลัมน์ผ่าน dictionary

การเลือกวิธีจัดการขึ้นอยู่กับบริบทของข้อมูล เช่น หากค่าว่างมีน้อยและไม่กระทบการวิเคราะห์อาจลบทิ้งได้ แต่หากข้อมูลมีค่าจำกัดควรเติมค่าแทนเพื่อไม่ให้เสียข้อมูลแถวอื่น ๆ ไป

```python
### Handling missing values ###

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, None, 22, 28],
    "Score": [85, 90, None, 92]
})
print(df)
print()

## Example 1: .isnull()
print(df.isna())
print()
print(df.isna().sum())
print()

## Example 2: .dropna()
df2 = df.dropna()
print(df2)
print()

df3 = df.dropna(subset=["Age"])
print(df3)
print()


## Example 3: .fillna()
df = df.fillna(0)
print(df)
print()

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Score"] = df["Score"].fillna(df["Score"].mean())
print(df)
print()

df2 = df.fillna({
    "Age": df["Age"].mean(),
    "Score": df["Score"].mean()
})
print(df2)
```

**คำอธิบาย:** การจัดการชื่อคอลัมน์และเลือกเก็บเฉพาะคอลัมน์ที่จำเป็นเป็นขั้นตอนพื้นฐานในการทำความสะอาดข้อมูล:
- `df.rename(columns={...})` - เปลี่ยนชื่อคอลัมน์เป็นชื่อใหม่ตาม dictionary ที่กำหนด (key คือชื่อเดิม, value คือชื่อใหม่)
- `df.drop(columns=[...])` - ลบคอลัมน์ที่ระบุออกจาก DataFrame หรือใช้ `df.drop("col", axis=1)` ก็ได้ผลลัพธ์เดียวกัน (`axis=1` หมายถึงดำเนินการตามแนวคอลัมน์)

ทั้งสองเมธอดคืนค่าเป็น DataFrame ใหม่โดยไม่แก้ไขต้นฉบับ เว้นแต่จะระบุ `inplace=True`

```python
### Renaming & dropping columns ###

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 21, 19, 22],
    "Math": [85, 90, 78, 92],
    "English": [88, 85, 80, 95]
})
print(df)

## Example 1: .rename()
df2 = df.rename(columns={
    "Name": "Student_Name",
    "Math": "Math_Score",
    "English": "English_Score"
})
print(df2)

## Example 2: .drop()
df3 = df.drop(columns=["Age", "English"])
print(df3)

df4 = df.drop("English", axis=1)
print(df4)
```

**คำอธิบาย:** `.replace()` ใช้แทนที่ **ค่าข้อมูล (values)** ภายในคอลัมน์ด้วยค่าใหม่ เช่น เปลี่ยน "F" เป็น "Female" โดยสามารถแทนที่ค่าเดียว หรือส่ง dictionary เพื่อแทนที่หลายค่าพร้อมกันในคำสั่งเดียว

ข้อควรระวังคือ `.replace()` ต่างจาก `.fillna()` ตรงที่ `.replace()` ใช้แทนที่ **ค่าที่มีอยู่จริง** (ต้องระบุค่าเดิมกับค่าใหม่) ในขณะที่ `.fillna()` ใช้เติมเฉพาะค่าที่เป็น **NaN/missing value** เท่านั้น (ระบุแค่ค่าที่จะเติม) จึงไม่สามารถใช้แทนกันได้

```python
### Replace Values ###

## Example 1: .replace()
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Gender": ["F", "M", "M", "F"],
    "Grade": ["A", "B", "A", "C"]
})
print(df)
print()

# df["Gender"] = df["Gender"].replace("F", "Female")
# print(df)

# df["Grade"] = df["Grade"].replace({
#     "A": "Excellent",
#     "B": "Good",
#     "C": "Average"
# })
# print(df)

df = df.replace({
    "F": "Female",
    "M": "Male"
})
print(df)

# Important .replace != .fillna()
# df["Age"] = df["Age"].replace(0, 25)          # Two parameters needed
# df["Age"] = df["Age"].fillna(25)              # Single parameter needed
```

**คำอธิบาย:** ข้อมูลซ้ำ (duplicate rows) อาจเกิดจากความผิดพลาดตอนรวบรวมข้อมูล ซึ่งควรตรวจสอบและจัดการก่อนวิเคราะห์:
- `df.duplicated()` - คืนค่า Series ของ True/False บอกว่าแต่ละแถวซ้ำกับแถวก่อนหน้าหรือไม่ (แถวที่พบครั้งแรกจะเป็น False เสมอ) ต่อด้วย `.sum()` เพื่อนับจำนวนแถวที่ซ้ำ
- `df.drop_duplicates()` - ลบแถวที่ซ้ำออก เหลือไว้เพียงแถวแรกที่พบ
- สามารถระบุ `subset=["column"]` เพื่อตรวจสอบความซ้ำเฉพาะบางคอลัมน์แทนที่จะเทียบทุกคอลัมน์

หลังจาก `drop_duplicates()` มักตามด้วย `.reset_index(drop=True)` เพื่อจัดเรียง index ใหม่ให้ต่อเนื่องอีกครั้ง

```python
### Detecting duplicates ###

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "Bob", "David", "Alice"],
    "Age": [20, 21, 22, 21, 23, 20],
    "Score": [85, 90, 78, 90, 92, 85]
})
print(df)
print()

## Example 1: .duplicate()
print(df.duplicated())
print(df.duplicated().sum())

duplicates = df[df.duplicated()]
print(duplicates)

df2 = df.drop_duplicates()
print(df2)

df2 = df.drop_duplicates().reset_index(drop=True)           # Drop duplicate usually use with reset index in practice
print(df2)

print(df.duplicated(subset=["Name"]))
```

**คำอธิบาย:** บางครั้งข้อมูลที่อ่านเข้ามาอาจมีชนิดข้อมูล (dtype) ไม่ตรงกับที่ต้องการ เช่น ตัวเลขถูกอ่านมาเป็นข้อความ (string) ซึ่งจะทำให้คำนวณทางคณิตศาสตร์ไม่ได้

`df["col"].astype(type)` ใช้แปลงชนิดข้อมูลของคอลัมน์ เช่น `astype(int)`, `astype(float)`, `astype(str)`, `astype(bool)` ควรตรวจสอบ `df.dtypes` ก่อนและหลังแปลงเสมอ เพื่อยืนยันว่าการแปลงชนิดข้อมูลสำเร็จตามที่ต้องการ

```python
### Type conversion ###

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": ["20", "21", "22"],
    "Score": ["85", "90", "78"],
    "Pass": [1, 1, 0]
})
print(df)
print(df.dtypes)

df["Name"] = df["Name"].astype(str)
df["Age"] = df["Age"].astype(int)
df["Score"] = df["Score"].astype(float)
df["Pass"] = df["Pass"].astype(bool)
print(df)
print(df.dtypes)
```

4. Transformation & Aggregation


**คำอธิบาย:** การประยุกต์ใช้ฟังก์ชันกับค่าข้อมูลใน pandas ทำได้ผ่าน `.map()` และ `.apply()` โดยหลักการคล้ายกันคือ นำแต่ละค่าในคอลัมน์ (Series) ไปผ่านฟังก์ชันที่กำหนด แล้วคืนค่าผลลัพธ์ออกมาเป็นคอลัมน์ใหม่

- `.map()` มักใช้กับ Series เพื่อแปลงค่าทีละตัว เช่นใช้ lambda function หรือ dictionary (แปลงค่าตามที่จับคู่ไว้ เช่น "Male" >>> 1)
- `.apply()` ก็ใช้กับ Series ได้เช่นกัน (และใช้กับทั้ง DataFrame เพื่อประยุกต์ฟังก์ชันตามแนวแถวหรือคอลัมน์ได้ด้วย)

ในตัวอย่างนี้ใช้ `.map()` ร่วมกับฟังก์ชันที่นิยามเอง (`add_bonus`) เพื่อบวกค่าคะแนนเพิ่มให้ทุกแถว

```python
### Applying functions ###

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Gender": ["Female", "Male", "Male"],
    "Age": [20, 25, 30],
    "Score": [80, 90, 75]
})
print(df)

# Example 1: .map()
# df["Score"] = df["Score"].map(lambda x: x + 1)      # Apply on each value
# print(df)

# df["Gender"] = df["Gender"].map({
#     "Male": 1,
#     "Female": 0
# })
# print(df)

# Example 2: .apply()
# df["Score"] = df["Score"].apply(lambda x: x + 5)
# print(df)

def add_bonus(x):
    return x + 5
df["Score"] = df["Score"].map(add_bonus)
print(df)
```

**คำอธิบาย:** `groupby()` ใช้จัดกลุ่มแถวข้อมูลตามค่าของคอลัมน์ที่กำหนด (เช่น จัดกลุ่มพนักงานตามแผนก) เพื่อคำนวณค่าสรุปในแต่ละกลุ่ม ข้อสำคัญคือ **`groupby()` เพียงอย่างเดียวไม่คำนวณอะไร** ต้องต่อด้วยฟังก์ชันรวมค่า (aggregation function) เสมอ เช่น
- `.mean()`, `.sum()` - คำนวณค่าเฉลี่ย/ผลรวมของคอลัมน์ที่เลือกในแต่ละกลุ่ม
- `.agg([...])` - คำนวณหลายค่าสถิติพร้อมกัน เช่น `['mean', 'max', 'min']`
- `.agg({...})` - กำหนดฟังก์ชันรวมค่าต่างกันสำหรับแต่ละคอลัมน์ผ่าน dictionary

ผลลัพธ์จะเป็น DataFrame หรือ Series ใหม่ที่มี index เป็นชื่อกลุ่ม (เช่นชื่อแผนก) แทนที่ index เดิม

```python
### GroupBy operations ###
data = {
    'Department': ['IT', 'IT', 'HR', 'HR', 'Sales', 'Sales', 'IT', 'Sales'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David',
                 'Eve', 'Frank', 'George', 'Helen'],
    'Salary': [50000, 60000, 45000, 48000, 55000, 65000, 52000, 70000],
    'Bonus': [5000, 6000, 4000, 4500, 5500, 6500, 5200, 7000]
}
df = pd.DataFrame(data)
print(df)

df.groupby('Department')['Salary'].mean()           # groupby() does not calculate anything by itself.
df.groupby('Department')['Salary'].sum()
df.groupby('Department')[['Salary', 'Bonus']].sum()
df.groupby('Department')['Salary'].agg(
    ['mean', 'max', 'min']
)
df.groupby('Department').agg({
    'Salary': ['mean', 'max'],
    'Bonus': ['sum', 'mean']
})
```

**คำอธิบาย:** Pivot table เป็นการสรุปข้อมูลในรูปแบบตารางไขว้ (cross-tabulation) คล้ายกับ Pivot Table ใน Excel โดยกำหนด 3 ส่วนหลักผ่าน `pd.pivot_table()`:
- **`values`** - คอลัมน์ที่ต้องการคำนวณ (What do I want to calculate?)
- **`index`** - คอลัมน์ที่ใช้จัดกลุ่มตามแถว (How do I group the rows?)
- **`aggfunc`** - ฟังก์ชันสรุปค่า เช่น `'mean'`, `'sum'` (How do I summarize?)

หากต้องการแยกผลลัพธ์ออกเป็นคอลัมน์ย่อยตามอีกหนึ่งมิติ (เช่นแยกเพศชาย/หญิงในแต่ละแผนก) ให้เพิ่มพารามิเตอร์ **`columns`** เข้าไป ผลลัพธ์ที่ได้จะเป็นตารางที่มีทั้งแถว (index) และคอลัมน์ (columns) แสดงค่าสรุปตามเงื่อนไขทั้งสองมิติพร้อมกัน

```python
### Pivot tables ###	
data = {
    'Department': ['IT', 'IT', 'HR', 'HR', 'Sales', 'Sales', 'IT', 'Sales'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David',
                 'Eve', 'Frank', 'George', 'Helen'],
    'Salary': [50000, 60000, 45000, 48000, 55000, 65000, 52000, 70000],
    'Bonus': [5000, 6000, 4000, 4500, 5500, 6500, 5200, 7000]
}
df = pd.DataFrame(data)
print(df)

# values   >>> What do I want to calculate?
# index    >>> How do I group the rows?
# aggfunc  >>> How do I summarize?
pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    aggfunc='mean'
)

# Columns  >>> How do I separate each value to single column?
df['Gender'] = [
    'F', 'M', 'M', 'F',
    'F', 'M', 'M', 'F'
]
print(df)
pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    columns='Gender',
    aggfunc='mean'
)
```

**คำอธิบาย:** `df.sort_values()` ใช้เรียงลำดับแถวข้อมูลตามค่าของคอลัมน์ที่ระบุ โดย default จะเรียงจากน้อยไปมาก (ascending) หากต้องการเรียงจากมากไปน้อยให้ระบุ `ascending=False`

สามารถเรียงตามหลายคอลัมน์พร้อมกันได้ผ่าน `by=[...]` และกำหนดทิศทางการเรียงของแต่ละคอลัมน์แยกกันผ่าน `ascending=[...]` **ข้อควรระวัง:** `sort_values()` จะไม่แก้ไข DataFrame เดิม แต่คืนค่าเป็น DataFrame ใหม่ที่เรียงลำดับแล้วเสมอ ต้องเก็บผลลัพธ์ไว้ในตัวแปรใหม่หากต้องการใช้งานต่อ นอกจากนี้ index ของแถวจะติดไปกับข้อมูลเดิม ไม่ได้เรียงใหม่ตามลำดับแถวที่แสดง

```python
### Sorting ###
data = {
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Department': ['IT', 'HR', 'Sales', 'IT', 'Sales'],
    'Salary': [60000, 45000, 70000, 52000, 55000]
}
df = pd.DataFrame(data)
print(df)


df.sort_values('Salary')                            # Sorting do not change order in original dataframe!
df.sort_values('Salary', ascending=False)

df.sort_values(
    by=['Department', 'Salary'],
    ascending=[True, False]
)

df2 = df.copy()
df2.index = [103, 101, 105, 102, 104]
print(df2)
```

5. Merging and Reshaping Data

**คำอธิบาย:** `pd.concat()` ใช้ต่อรวม DataFrame ตั้งแต่ 2 ตัวขึ้นไปเข้าด้วยกัน โดยกำหนดทิศทางผ่านพารามิเตอร์ `axis`:
- `axis=0` (default) - ต่อรวมตามแนว **แถว** (เพิ่มจำนวนแถว) เหมาะกับข้อมูลที่มีโครงสร้างคอลัมน์เหมือนกัน ควรระบุ `ignore_index=True` เพื่อรีเซ็ต index ใหม่ให้ต่อเนื่อง
- `axis=1` - ต่อรวมตามแนว **คอลัมน์** (เพิ่มจำนวนคอลัมน์) โดยจับคู่แถวตาม index เดิม

หาก DataFrame ที่นำมาต่อกันมีคอลัมน์ไม่ตรงกันทั้งหมด (เช่นตัวอย่างที่ 3) ตำแหน่งที่ไม่มีข้อมูลตรงกันจะถูกเติมด้วย `NaN` โดยอัตโนมัติ

```python
### Concatenation ###

## Example 1: Extend Rows
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['John', 'Jane', 'Bob'],
    'Score': [80, 75, 90]
})
df2 = pd.DataFrame({
    'ID': [4, 5, 6],
    'Name': ['Alice', 'David', 'Emma'],
    'Score': [85, 70, 95]
})

df = pd.concat([df1, df2], axis=0)                              # axis=0 is default, can omit it
df = pd.concat([df1, df2], axis=0, ignore_index=True)           # Recommend to reset index 
print(df)



## Example 2 : Extend columns
name = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob']
})
score = pd.DataFrame({
    'Score': [80, 75, 90]
})
df = pd.concat([name, score], axis=1)                           # axis=1 cannot omit for extending columns
print(df)



## Example 3: Concat different dataframe
df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['John', 'Jane']
})
df2 = pd.DataFrame({
    'ID': [3, 4],
    'Score': [90, 80]
})
result = pd.concat([df1, df2], ignore_index=True)
print(result)
```

**คำอธิบาย:** การรวมข้อมูลจากหลายตารางเข้าด้วยกันโดยอ้างอิงคีย์ร่วม (เหมือน JOIN ใน SQL) ทำได้ 2 แบบหลัก:
- **`pd.merge()`** - ใช้เมื่อคีย์ที่จะจับคู่อยู่ใน **คอลัมน์** ของทั้งสองตาราง ระบุคอลัมน์ที่ใช้จับคู่ผ่าน `on="column"`
- **`df.join()`** - ใช้เมื่อคีย์ที่จะจับคู่อยู่ใน **index** ของทั้งสองตาราง

ทั้งสองเมธอดกำหนดรูปแบบการรวมผ่านพารามิเตอร์ `how` ได้เหมือนกัน:
- `'inner'` - เก็บเฉพาะคีย์ที่มีอยู่ในทั้งสองตาราง
- `'outer'` - เก็บคีย์ทั้งหมดจากทั้งสองตาราง (ตำแหน่งที่ไม่มีข้อมูลจะเป็น `NaN`)
- `'left'` - ยึดคีย์จากตารางซ้ายเป็นหลัก
- `'right'` - ยึดคีย์จากตารางขวาเป็นหลัก

```python
### Merging/Joining ###

##Example 1: Merge() 
# students = pd.DataFrame({
#     'student_id': [101, 102, 103, 104],
#     'name': ['Alice', 'Bob', 'Charlie', 'David']
# })
# scores = pd.DataFrame({
#     'student_id': [101, 102, 103, 105],
#     'score': [85, 72, 90, 65]
# })

# result = pd.merge(students, scores, on='student_id', how='inner')             # If key is in column
# result = pd.merge(students, scores, on='student_id', how='outer')
# result = pd.merge(students, scores, on='student_id', how='left')
# result = pd.merge(students, scores, on='student_id', how='right')
# print(result)


##Example 2: Join() 
students = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'David']
}, index=[101, 102, 103, 104])

scores = pd.DataFrame({
    'score': [85, 72, 90, 65]
}, index=[101, 102, 103, 105])

result = students.join(scores)                                                  # If key is in index
result = students.join(scores, how='right')                     
result = students.join(scores, how='inner')
result = students.join(scores, how='outer')
print(result)
```

**คำอธิบาย:** การแปลงรูปแบบตาราง (reshaping) เป็นการจัดโครงสร้างข้อมูลใหม่โดยไม่เปลี่ยนเนื้อหา ใช้บ่อยเมื่อต้องเตรียมข้อมูล ให้อยู่ในรูปแบบที่เหมาะกับการวิเคราะห์หรือพล็อตกราฟ:
- **`.melt()`** - แปลงข้อมูลจากรูปแบบ "wide" (แต่ละตัวแปรเป็นคอลัมน์) ให้เป็นรูปแบบ "long" (รวมคอลัมน์ตัวแปรเป็นแถวเดียว) โดยระบุ `id_vars` (คอลัมน์ที่คงไว้), `var_name` (ชื่อคอลัมน์ใหม่ที่เก็บชื่อตัวแปรเดิม) และ `value_name` (ชื่อคอลัมน์ใหม่ที่เก็บค่า)
- **`.stack()`** - แปลงคอลัมน์ให้กลายเป็นระดับหนึ่งของ index (บีบข้อมูลจากแนวนอนเป็นแนวตั้ง)
- **`.unstack()`** - ทำสิ่งตรงข้ามกับ `.stack()` คือแปลง index กลับมาเป็นคอลัมน์

```python
### Reshaping ###

## Example 1: .melt()
# df = pd.DataFrame({
#     'Name': ['John', 'Jane', 'Bob'],
#     'Math': [80, 75, 90],
#     'English': [85, 80, 88],
#     'Science': [90, 85, 92]
# })
# df_long = df.melt(
#     id_vars='Name',                               # Remaining column 
#     var_name='Subject',                           # New column name of original columns
#     value_name='Score'                            # New column for the data
# )
# print(df_long)



# Example 2: .stack() and .unstack
df = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob'],
    'Math': [80, 75, 90],
    'English': [85, 80, 88],
    'Science': [90, 85, 92]
})
df = df.set_index('Name')
print(df)

result = df.stack()                                # Columns convert to Index 
print(result)

result = result.unstack()                           # Index convert to Columns
print(result)
```
