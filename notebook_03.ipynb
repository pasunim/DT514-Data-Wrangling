### 9. Select Rows Using .loc[ ]
Use .loc[ ] to select records with index values from 100 to 109 and display only the following columns: "No", "Age", "Income"

**คำอธิบาย:** ข้อนี้ใช้ `.loc[ ]` ซึ่งเป็นการเลือกข้อมูลโดยอ้างอิงจาก "label" ของ index และชื่อคอลัมน์ (ต่างจาก `.iloc[ ]` ที่อ้างอิงตามตำแหน่งตัวเลข) รูปแบบคือ `df.loc[แถวที่ต้องการ, คอลัมน์ที่ต้องการ]`

`df.loc[100:109, ["No", "Age", "Income"]]` หมายถึง:
- เลือกแถวที่มีค่า index ตั้งแต่ 100 ถึง 109 โดยข้อควรระวังคือ `.loc[]` จะ "รวม" ค่าขอบขวาไว้ด้วย (inclusive) ต่างจากการ slice แบบปกติของ Python ดังนั้นจะได้ทั้งหมด 10 แถว (100, 101, ..., 109)
- เลือกเฉพาะคอลัมน์ "No", "Age", "Income" โดยระบุเป็น list ของชื่อคอลัมน์

วิธีนี้เหมาะกับการเลือกข้อมูลเมื่อทราบชื่อคอลัมน์หรือค่า index ที่ต้องการอยู่แล้ว

```python
df.loc[100:109, ["No", "Age", "Income"]]
```

### 10. Select Rows Using .iloc[ ]
Use .iloc[ ] to select records 100–109, columns 1–5

**คำอธิบาย:** ข้อนี้ใช้ `.iloc[ ]` ซึ่งต่างจาก `.loc[ ]` ตรงที่ `.iloc[ ]` อ้างอิงตำแหน่ง (position) แบบตัวเลขล้วน ๆ ไม่สนใจค่า label ของ index หรือชื่อคอลัมน์ โดยรูปแบบคือ `df.iloc[แถวเริ่ม:แถวจบ, คอลัมน์เริ่ม:คอลัมน์จบ]`

`df.iloc[100:110, 1:6]` หมายถึง:
- แถว: ตำแหน่งที่ 100 ถึง 109 (ไม่รวมตำแหน่งที่ 110 เพราะการ slice ด้วยตัวเลขใน iloc ไม่รวมขอบขวา เหมือนการ slice list ทั่วไปของ Python) รวมเป็น 10 แถว
- คอลัมน์: ตำแหน่งที่ 1 ถึง 5 (ไม่รวมตำแหน่งที่ 6) คือคอลัมน์ Income, Kidhome, Teenhome, Recency, MntWines (นับตำแหน่ง 0 คือ No)

ข้อแตกต่างสำคัญระหว่าง `.loc[100:109]` (ข้อ 9) กับ `.iloc[100:110]` (ข้อนี้) คือ `.loc[]` รวมขอบขวา (100 ถึง 109 รวม 109) ในขณะที่ `.iloc[]` ไม่รวมขอบขวา จึงต้องใช้ 110 เพื่อให้ได้แถวสุดท้ายคือตำแหน่ง 109 เช่นกัน

```python
df.iloc[100:110, 1:6]
```

### 11. Filter by Age
Select only customers who are older than 40 years. Display: "No", "Age", "Income", "Response".

**คำอธิบาย:** ข้อนี้ต้องการกรอง (filter) แถวข้อมูลตามเงื่อนไขที่กำหนด โดยใช้ `df.loc[เงื่อนไข, [คอลัมน์ที่ต้องการ]]`
- เงื่อนไข `df["Age"] > 40` จะสร้าง Series ของค่า True/False ที่บอกว่าลูกค้าแต่ละแถวมีอายุมากกว่า 40 ปีหรือไม่
- เมื่อนำ Series นี้ไปใส่ใน `.loc[]` pandas จะเลือกเฉพาะแถวที่มีค่าเป็น True เท่านั้น
- ส่วนที่สองใน `.loc[]` คือ list ของชื่อคอลัมน์ `["No", "Age", "Income", "Response"]` ซึ่งใช้ระบุว่าต้องการแสดงเฉพาะคอลัมน์เหล่านี้

ผลลัพธ์ที่ได้คือ DataFrame ที่มีเฉพาะลูกค้าอายุมากกว่า 40 ปี และแสดงเพียง 4 คอลัมน์ตามที่โจทย์กำหนด ถือเป็นรูปแบบพื้นฐานของการกรองข้อมูลแบบมีเงื่อนไข (conditional filtering) ที่ใช้บ่อยมากในการวิเคราะห์ข้อมูล

```python
df.loc[df["Age"] > 40, ["No", "Age", "Income", "Response"]]
```

### 13. Filter Using Multiple Conditions
Select only customers who meet both conditions: "Age" > 40, "Income" > 50,000

ใช้รูปแบบ `df[(เงื่อนไข1) & (เงื่อนไข2)]` โดยต้องครอบแต่ละเงื่อนไขด้วยวงเล็บเสมอ เพราะตัวดำเนินการ `&` ใน pandas มีลำดับความสำคัญ (operator precedence) สูงกว่าตัวเปรียบเทียบ (`>`, `<`, `==`) หากไม่ใส่วงเล็บจะเกิด error ผลลัพธ์ที่ได้จะมีเฉพาะแถวที่ผ่าน "ทั้งสองเงื่อนไข" พร้อมกันเท่านั้น (แคบกว่าการกรองด้วยเงื่อนไขเดียว)

Remove records containing missing values and compare the number of records:

### 20. Fill Missing Values
Create a new DataFrame and replace missing values in `Income` with the mean value of `Income`.

**คำอธิบาย:** ข้อนี้ต้องการเติมค่าที่ขาดหาย (missing values) ในคอลัมน์ "Income" แทนที่จะลบแถวนั้นทิ้งไปเลยเหมือนข้อ 19 โดยมีขั้นตอนดังนี้:
1. คัดลอก `df_missing` มาเป็น `df_filled` ด้วย `.copy()` เพื่อไม่ให้กระทบข้อมูลต้นฉบับ
2. คำนวณค่าเฉลี่ย (mean) ของคอลัมน์ Income จากค่าที่ไม่ขาดหาย ด้วย `df_filled["Income"].mean()` (pandas จะข้าม NaN โดยอัตโนมัติเมื่อคำนวณ mean)
3. ใช้ `.fillna(income_mean)` เพื่อแทนที่ค่า NaN ทุกตำแหน่งในคอลัมน์ Income ด้วยค่าเฉลี่ยที่คำนวณได้
4. ตรวจสอบผลลัพธ์ว่าไม่มีค่า NaN เหลืออยู่ในคอลัมน์ Income แล้ว ด้วย `.isnull().sum()`

การเติมค่าด้วยค่าเฉลี่ย (mean imputation) เป็นวิธีพื้นฐานที่ช่วยรักษาจำนวนแถวข้อมูลไว้ครบ แทนที่จะสูญเสียข้อมูลจากการลบแถวทิ้ง ซึ่งเหมาะกับกรณีที่สัดส่วนข้อมูลขาดหายไม่มากและคอลัมน์นั้นมีการกระจายตัวไม่เบ้มากนัก

```python
df_filled = df_missing.copy()

income_mean = df_filled["Income"].mean()
df_filled["Income"] = df_filled["Income"].fillna(income_mean)

print("Mean Income used to fill:", income_mean)
print("Missing values in Income after filling:", df_filled["Income"].isnull().sum())
df_filled
```

### 24. Detect Duplicates
Check whether the dataset contains duplicate records. Report: (A) Number of duplicate records, (B) Duplicate records found

Note: `df` has no duplicate records, so a few duplicate rows are appended here (`df_dup`) to demonstrate duplicate detection.

**คำอธิบาย:** ข้อนี้ต้องการตรวจสอบว่าข้อมูลมีแถวที่ซ้ำกันทุกคอลัมน์หรือไม่ โดยใช้เมธอด `df.duplicated()` ซึ่งจะคืนค่า True/False สำหรับแต่ละแถว (True หมายถึงแถวนั้นซ้ำกับแถวก่อนหน้า) แล้วนำมา `.sum()` เพื่อนับจำนวนแถวที่ซ้ำทั้งหมด

เนื่องจากข้อมูลต้นฉบับ `df` ไม่มีแถวซ้ำอยู่แล้ว (ผลลัพธ์เป็น 0) จึงจำลองสถานการณ์โดยสร้าง `df_dup` ที่คัดลอกแถวแรก ๆ (index 0, 1, 2) มาต่อท้ายข้อมูลเดิมด้วย `pd.concat()` เพื่อให้เห็นตัวอย่างการตรวจจับข้อมูลซ้ำจริง ๆ จากนั้น:
- (A) นับจำนวนแถวซ้ำด้วย `.duplicated().sum()`
- (B) แสดงแถวที่ซ้ำทั้งหมดด้วย `.duplicated(keep=False)` ซึ่งจะทำเครื่องหมาย True ทั้งแถวต้นฉบับและแถวที่คัดลอกมา (ต่างจากค่า default ที่จะมองว่ามีแค่แถวหลังเป็นแถวซ้ำ)

Create a new column named "Response_Label" based on "Response": 1 >>> Accepted, 0 >>> Not Accepted

### 28. Create a new column named "Income_Level"
Divide customers into three groups: (1) Income < 30,000 >>> Low, (2) 30,000–60,000 >>> Medium, (3) Income > 60,000 >>> High

**คำอธิบาย:** ข้อนี้ต้องการแบ่งกลุ่มลูกค้าตามระดับรายได้ (Income) ออกเป็น 3 กลุ่ม โดยสร้างฟังก์ชัน `income_level()` ที่รับค่ารายได้เข้ามาแล้วตรวจสอบเงื่อนไขทีละขั้น (if-elif-else):
- ถ้า Income < 30,000 >>> คืนค่า "Low"
- ถ้า Income อยู่ระหว่าง 30,000–60,000 (รวมถึง 60,000) >>> คืนค่า "Medium"
- นอกเหนือจากนั้น (มากกว่า 60,000) >>> คืนค่า "High"

จากนั้นใช้ `df["Income"].apply(income_level)` เพื่อนำฟังก์ชันนี้ไปประมวลผลกับค่า Income ของลูกค้าทุกแถว แล้วเก็บผลลัพธ์ไว้ในคอลัมน์ใหม่ชื่อ "Income_Level" วิธีนี้เหมาะกับกรณีที่เงื่อนไขการแบ่งกลุ่มมีหลายขั้นตอนหรือซับซ้อนกว่าการเปรียบเทียบแบบง่าย ๆ

```python
def income_level(income):
    if income < 30000:
        return "Low"
    elif income <= 60000:
        return "Medium"
    else:
        return "High"

df["Income_Level"] = df["Income"].apply(income_level)

df[["No", "Income", "Income_Level"]]
```

### 29. Customer Analysis 1
Group customers by "Education" and calculate: (1) Number of customers, (2) Average Income, (3) Average Age, (4) Average Wine Spending, (5) Average Web Purchases, (6) Average Store Purchases

**คำอธิบาย:** ข้อนี้ต้องการวิเคราะห์ข้อมูลลูกค้าโดยแบ่งกลุ่มตามระดับการศึกษา (Education) แล้วสรุปค่าสถิติของแต่ละกลุ่ม ใช้เมธอด `df.groupby("Education").agg(...)` ซึ่งเป็นการทำงานแบบ "Split-Apply-Combine":
1. **Split** - แบ่งข้อมูลออกเป็นกลุ่มย่อยตามค่าที่ไม่ซ้ำกันในคอลัมน์ Education (เช่น Graduate, Doctorate, Master, Basic, 2nCycle)
2. **Apply** - คำนวณค่าสถิติที่ต้องการในแต่ละกลุ่ม ได้แก่ จำนวนลูกค้า (`count`) และค่าเฉลี่ย (`mean`) ของ Income, Age, MntWines, NumWebPurchases, NumStorePurchases
3. **Combine** - รวมผลลัพธ์ของแต่ละกลุ่มกลับมาเป็นตารางเดียว โดยแต่ละแถวคือระดับการศึกษาหนึ่งกลุ่ม

การตั้งชื่อในรูปแบบ `Average_Income=("Income", "mean")` คือการใช้ "named aggregation" ของ pandas ทำให้ได้ชื่อคอลัมน์ผลลัพธ์ที่สื่อความหมายชัดเจน แทนที่จะได้ชื่อคอลัมน์เดิมซ้ำกัน

```python
education_summary = df.groupby("Education").agg(
    Number_of_Customers=("No", "count"),
    Average_Income=("Income", "mean"),
    Average_Age=("Age", "mean"),
    Average_Wine_Spending=("MntWines", "mean"),
    Average_Web_Purchases=("NumWebPurchases", "mean"),
    Average_Store_Purchases=("NumStorePurchases", "mean"),
)

education_summary
```

### 30. Customer Analysis 2
Create a Pivot Table with: Index = Education and Columns = Response and Values = Number of customer

**คำอธิบาย:** ข้อนี้ต้องการสร้าง Pivot Table เพื่อดูภาพรวมความสัมพันธ์ระหว่างระดับการศึกษา (Education) กับการตอบรับแคมเปญ (Response) โดยใช้เมธอด `df.pivot_table()`
- `index="Education"` >>> กำหนดให้ค่าของคอลัมน์ Education เป็นแถว (แกนตั้ง) ของตาราง
- `columns="Response"` >>> กำหนดให้ค่าของ Response (0 หรือ 1) เป็นคอลัมน์ (แกนนอน) ของตาราง
- `values="No"` และ `aggfunc="count"` >>> นับจำนวนลูกค้า (นับจากคอลัมน์ No ซึ่งไม่ซ้ำกัน) ที่ตกอยู่ในแต่ละช่อง (Education × Response)

ผลลัพธ์ที่ได้คือตารางไขว้ (cross-tabulation) ที่บอกว่าลูกค้าแต่ละระดับการศึกษา มีจำนวนกี่คนที่ "ตอบรับ" (Response = 1) และกี่คนที่ "ไม่ตอบรับ" (Response = 0) แคมเปญล่าสุด ทำให้เห็นได้ทันทีว่ากลุ่มการศึกษาใดตอบรับแคมเปญในสัดส่วนที่สูงกว่ากลุ่มอื่น

```python
pivot_table = df.pivot_table(
    index="Education",
    columns="Response",
    values="No",
    aggfunc="count",
)

pivot_table
```
